## 3. Príprava dát


### 3.1. Výber a rozdelenie dát

```python
# Oddelenie cieľového atribútu a vstupných príznakov

y = data_hr["Attrition"].map({"No": 0, "Yes": 1})

X = data_hr.drop(columns=["Attrition", "EmployeeNumber"])
```

```python
# Stratifikované rozdelenie dát na trénovaciu a testovaciu množinu

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, stratify=y, random_state=42)
```

```python
# Odstránenie konštantných atribútov

constant_cols = X_train.columns[X_train.nunique() <= 1]

X_train = X_train.drop(columns=constant_cols)
X_test = X_test.drop(columns=constant_cols)
```

### 3.2. Výber relevantných atribútov

```python
# Vyhodnotenie súvislosti číselných atribútov pomocou Spearmanovej korelácie

numeric_ranking = (
    X_train.select_dtypes(include="number")
    .corrwith(y_train, method="spearman")
    .sort_values(key=lambda values: values.abs(), ascending=False)
)

display(numeric_ranking)
```

```python
# Vyhodnotenie súvislosti kategorických atribútov pomocou Cramérovho V

from scipy.stats.contingency import association

category_cols = X_train.select_dtypes(include="object").columns

category_scores = {}

for col in category_cols:
    table = pd.crosstab(X_train[col], y_train)
    category_scores[col] = association(table, method="cramer")

category_ranking = pd.Series(category_scores).sort_values(ascending=False)

display(category_ranking)
```

```python
# Výber atribútov podľa stanovenej prahovej hodnoty

threshold = 0.05

selected_numeric = numeric_ranking[numeric_ranking.abs() > threshold].index.tolist()

selected_categories = category_ranking[category_ranking > threshold].index.tolist()

selected_cols = selected_numeric + selected_categories

X_train_selected = X_train[selected_cols]
X_test_selected = X_test[selected_cols]

print(f"Selected attributes: {len(selected_cols)}")
print(selected_cols)
```

### 3.3. Transformácia dát

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

number_cols = X_train_selected.select_dtypes(include="number").columns
text_cols = X_train_selected.select_dtypes(include="object").columns

# Alternatívny experiment so všetkými vstupnými atribútmi

# number_cols = X_train.select_dtypes(include="number").columns
# text_cols = X_train.select_dtypes(include="object").columns

preprocessor = ColumnTransformer([
    ("numbers", StandardScaler(), number_cols),
    ("categories", OneHotEncoder(handle_unknown="ignore",
                                 sparse_output=False), text_cols)
])
```

```python
# Transformácie sa učia iba na trénovacích dátach

X_train_ready = preprocessor.fit_transform(X_train_selected)
X_test_ready = preprocessor.transform(X_test_selected)
```

### Modelovanie

```python
from sklearn.metrics import ConfusionMatrixDisplay, RocCurveDisplay
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
)


def evaluate_model(model, X, y):
    y_pred = model.predict(X)
    y_prob = model.predict_proba(X)[:, 1]

    fig, (ax_cm, ax_roc) = plt.subplots(1, 2, figsize=(10, 4))

    ConfusionMatrixDisplay.from_predictions(
        y, y_pred, display_labels=["Stayed", "Left"],
        cmap="Blues", colorbar=False, ax=ax_cm,
    )
    ax_cm.set(title="Confusion matrix", xlabel="Predicted", ylabel="Actual")
    ax_cm.grid(False)

    RocCurveDisplay.from_predictions(y, y_prob, ax=ax_roc)
    ax_roc.plot([0, 1], [0, 1], "--", color="gray", label="Random (AUC = 0.50)")
    ax_roc.set(title="ROC curve", xlabel="False positive rate", ylabel="Recall")
    ax_roc.legend(loc="lower right")

    fig.tight_layout()
    plt.show()

    return {
        "Accuracy": accuracy_score(y, y_pred),
        "Precision": precision_score(y, y_pred, zero_division=0),
        "Recall": recall_score(y, y_pred, zero_division=0),
        "F1": f1_score(y, y_pred, zero_division=0),
        "ROC-AUC": roc_auc_score(y, y_prob),
    }
```

```python
def recall_at_20(model, X, y):
    scores = model.predict_proba(X)[:, 1]
    y = np.asarray(y)

    budget = len(y) // 5
    top = np.argsort(-scores, kind="stable")[:budget]

    return (y[top] == 1).sum() / (y == 1).sum()
```

```python
cluster_cols = [
    "Age",
    "MonthlyIncome",
    "YearsAtCompany",
    "DistanceFromHome"
]
```

```python
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score

cluster_data = X_train[cluster_cols].copy()

cluster_scaler = StandardScaler()
Z = cluster_scaler.fit_transform(cluster_data)
```

```python
cluster_results = []
cluster_labels = {}

for k in range(2, 7):
    models = {
        "K-means": KMeans(
            n_clusters=k,
            n_init=20,
            random_state=42
        ),
        "Hierarchical": AgglomerativeClustering(
            n_clusters=k,
            linkage="ward"
        )
    }

    for name, model in models.items():
        labels = model.fit_predict(Z)

        cluster_labels[(name, k)] = labels
        cluster_results.append({
            "Method": name,
            "Clusters": k,
            "Silhouette": silhouette_score(Z, labels)
        })

cluster_scores = pd.DataFrame(cluster_results)
display(cluster_scores.round(3))
```

```python
sns.lineplot(
    data=cluster_scores,
    x="Clusters",
    y="Silhouette",
    hue="Method",
    marker="o"
)

plt.xticks(range(2, 7))
plt.title("Výber počtu zhlukov")
plt.ylabel("Silhouette score")
plt.xlabel("Počet zhlukov")
plt.tight_layout()
plt.show()
```

```python
best_clusters = cluster_scores.loc[
    cluster_scores.groupby("Method")["Silhouette"].idxmax()
]

display(best_clusters.round(3))
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
coordinates = pca.fit_transform(Z)

print(
    "Доля дисперсии на графике:",
    round(pca.explained_variance_ratio_.sum(), 3)
)
```

```python
def describe_clusters(name, labels):
    groups = labels + 1

    data = cluster_data.copy()
    data["Cluster"] = groups
    data["Attrition"] = y_train

    grouped = data.groupby("Cluster")

    summary = grouped[cluster_cols].mean()
    summary.insert(0, "Count", grouped.size())
    summary["Left_count"] = grouped["Attrition"].sum()
    summary["Attrition_%"] = grouped["Attrition"].mean() * 100

    print(name)
    display(summary.round(2))

    profiles = pd.DataFrame(Z, columns=cluster_cols)
    profiles["Cluster"] = groups
    profiles = profiles.groupby("Cluster").mean()

    fig, axes = plt.subplots(1, 3, figsize=(17, 4))

    sns.heatmap(
        profiles, annot=True, fmt=".2f",
        cmap="coolwarm", center=0, ax=axes[0]
    )
    axes[0].set_title("Profily zhlukov")
    axes[0].tick_params(axis="x", rotation=45)

    sns.scatterplot(
        x=coordinates[:, 0],
        y=coordinates[:, 1],
        hue=groups.astype(str),
        palette="tab10",
        alpha=0.6,
        s=25,
        ax=axes[1]
    )
    axes[1].set(
        title="PCA projekcia",
        xlabel="PC1", ylabel="PC2"
    )
    axes[1].legend(title="Zhluk")

    axes[2].bar(
        summary.index.astype(str),
        summary["Attrition_%"],
        color="steelblue"
    )
    axes[2].axhline(
        y_train.mean() * 100,
        color="red", linestyle="--",
        label="Celkový podiel odchodov"
    )
    axes[2].set(
        title="Odchody podľa zhlukov",
        xlabel="Zhluk",
        ylabel="Podiel odchodov (%)"
    )
    axes[2].legend()

    fig.suptitle(name)
    plt.tight_layout()
    plt.show()

    return summary
```

```python
cluster_summaries = {}

for _, row in best_clusters.iterrows():
    name = row["Method"]
    k = int(row["Clusters"])

    cluster_summaries[name] = describe_clusters(
        name,
        cluster_labels[(name, k)]
    )
```
